```python
import cooler
import pandas as pd

# Load the cooler file
#c = cooler.Cooler('/data/processing4/erikson_pipegrp/Jumana_nextflow/20250325_merged.mcool::/resolutions/10000')
resolution = 10000 # note this might be slightly slow on a laptop
                  # and could be lowered to 10kb for increased speed
clr = cooler.Cooler('/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004895_PnM1/pairs/20250325_24L004895_PnM1.pairs.dedup.cool')


# Specify the chromosomes
chrom1 = '2L_JJg14_057'
chrom2 = '2L_JJg14_439'

# Extract trans contacts between the specified chromosomes
trans_contacts = clr.matrix(as_pixels=True, balance=False).fetch((chrom1, chrom2))
print(trans_contacts.head())

# Create a DataFrame for the new cooler file
df = pd.DataFrame({
    'bin1_id': trans_contacts['bin1_id'],
    'bin2_id': trans_contacts['bin2_id'],
    'count': trans_contacts['count']
})

# Use the bins from the original cooler object
bins = clr.bins()[:]  # Extract bins from the original cooler file

# Create a new cooler file with the extracted trans contacts
cooler.create_cooler(
    '/data/processing4/erikson_pipegrp/Jumana_nextflow/20250325_24L004895_PnM1.2L.cool',
    bins=bins,
    pixels=df,
    ordered=True,
    dtypes={'count': 'int32'}  # Ensure the data type is specified for the count
)
```

```python
#Filtering the file into maternal and paternal chromosome files 439

import sys
import gzip

def filter_pairs_by_chromosome(input_file, output_file):
    pattern = "JJg14_439"
    buffer = []
    buffer_size = 1000
    with gzip.open(input_file, 'rt') as infile, open(output_file, 'w') as outfile:
        for line in infile:
            if line.startswith("#"):  # Process header lines
                if "_JJg14_057" in line:
                    continue
                line = line.replace("_JJg14_439", "")
                buffer.append(line)
                if len(buffer) >= buffer_size:
                    outfile.writelines(buffer)
                    buffer = []
                continue
            parts = line.strip().split("\t")
            if pattern in parts[1] and pattern in parts[3]:  # Check both chromosomes
                parts[1] = parts[1].replace("_JJg14_439", "")
                parts[3] = parts[3].replace("_JJg14_439", "")
                buffer.append("\t".join(parts) + "\n")
                if len(buffer) >= buffer_size:
                    outfile.writelines(buffer)
                    buffer = []
        if buffer:
            outfile.writelines(buffer)

#nput_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/24L004896_PnM2.pairs.gz"
#output_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250331_24L004896_PnM2.439.pairs"
input_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004897_PnM3/pairs/24L004897_PnM3.pairs.gz"
output_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004897_PnM3/pairs/20250331_24L004897_PnM3.439.pairs"
filter_pairs_by_chromosome(input_file, output_file)

```

```python
import sys
import gzip

def filter_pairs_by_specific_chromosomes(input_file, output_file):
    # List of chromosomes to filter
    target_chromosomes = ["2L_JJg14_439","2R_JJg14_439", "3L_JJg14_439", "3R_JJg14_439", "4_JJg14_439", "X_JJg14_439", "Y_JJg14_439", "2L_JJg14_057", "2R_JJg14_057", "3L_JJg14_057", "3R_JJg14_057", "4_JJg14_057", "X_JJg14_057", "Y_JJg14_057"]
    buffer = []
    buffer_size = 10000

    with gzip.open(input_file, 'rt') as infile, open(output_file, 'w') as outfile:
        for line in infile:
            if line.startswith("#"):  # Process header lines
                # Check if the header line starts with any of the target chromosomes
                if line.startswith("#chromsize"):
                    parts = line.strip().split(" ")
                    if parts[1] in target_chromosomes:
                        buffer.append(line)
                elif line.startswith("#samheader: @SQ"):
                    parts = line.strip().split(":")
                    parts2 = parts[2].strip().split("LN")
                    if parts2[0].strip() in target_chromosomes:
                        buffer.append(line)
                else:
                    buffer.append(line)
                if len(buffer) >= buffer_size:
                    outfile.writelines(buffer)
                    buffer = []
                continue
            parts = line.strip().split("\t")
            # Check if both parts[1] and parts[3] are in the target chromosomes list
            if parts[1] in target_chromosomes and parts[3] in target_chromosomes:
                buffer.append("\t".join(parts) + "\n")
                if len(buffer) >= buffer_size:
                    outfile.writelines(buffer)
                    buffer = []
        if buffer:
            outfile.writelines(buffer)

# Example usage
#input_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/24L004896_PnM2.pairs.gz"
#output_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250403_24L004896_PnM2.filtered.pairs"

#input_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004895_PnM1/pairs/24L004895_PnM1.pairs.gz"
#output_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004895_PnM1/pairs/20250403_24L004895_PnM1.filtered.pairs"

input_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004897_PnM3/pairs/24L004897_PnM3.pairs.gz"
output_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004897_PnM3/pairs/20250403_24L004897_PnM3.filtered.pairs"


filter_pairs_by_specific_chromosomes(input_file, output_file)
```

```python
#Extract Trans-hom (Trans-homologous):
#This term refers to interactions between homologous chromosomes. Homologous chromosomes are pairs of chromosomes that have the same genes at the same loci but may have different alleles. In diploid organisms, one chromosome of each homologous pair is inherited from each parent.

import sys
import gzip

def filter_by_cis_trans(input_file,output_file_cis, output_file_trans):
    
    buffer = []
    buffer_size = 10000
    buffer_trans= []
    buffer_trans_size = 10000

    with gzip.open(input_file, 'rt') as infile, open(output_file_cis, 'w') as output_cis, open(output_file_trans, 'w') as output_trans:
        for line in infile:
            if line.startswith("#"):
                buffer.append(line)
                buffer_trans.append(line)
            else:
                parts = line.strip().split("\t")
                if parts[1] == parts[3]:
                    buffer.append(line)
                    if len(buffer) >= buffer_size:
                        output_cis.writelines(buffer)
                        buffer = []        
                else:
                    buffer_trans.append(line)
                    if len(buffer_trans) >= buffer_trans_size:
                        output_trans.writelines(buffer_trans)
                        buffer_trans =[]
                
        if buffer:
            output_cis.writelines(buffer)
        
        if buffer_trans:
            output_trans.writelines(buffer_trans)

                

input_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004895_PnM1/pairs/20250403_24L004895_PnM1.filtered.pairs.gz"
output_file_cis = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004895_PnM1/pairs/20250501_24L004895_PnM1.cis.filtered.pairs"
output_file_trans = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004895_PnM1/pairs/20250501_24L004895_PnM1.trans.filtered.pairs"

#input_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250403_24L004896_PnM2.filtered.pairs.gz"
#output_file_cis = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250501_24L004896_PnM2.cis.filtered.pairs"
#output_file_trans = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250501_24L004896_PnM2.trans.filtered.pairs"

#input_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004897_PnM3/pairs/20250403_24L004897_PnM3.filtered.pairs.gz"
#output_file_cis = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004897_PnM3/pairs/20250501_24L004897_PnM3.cis.filtered.pairs"
#output_file_trans = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004897_PnM3/pairs/20250501_24L004897_PnM3.trans.filtered.pairs"

filter_by_cis_trans(input_file,output_file_cis, output_file_trans)
```

```python
#Filtering the trans interactions and renaming them to only the chromosome name
import sys
import gzip

def filter_by_cis_trans(input_file,output_file_cis, output_file_trans_hom, output_file_trans_het):
    
    buffer = []
    buffer_size = 10000

    with gzip.open(input_file, 'rt') as infile, open(output_file_cis, 'w') as output_cis, open(output_file_trans_hom , 'w') as output_trans_hom, open(output_file_trans_het, 'w') as output_trans_het:
        for line in infile:
            if line.startswith("#"):
                output_cis.writelines(line)
                cleaned_line = line.replace("_JJg14_057", "").replace("_JJg14_439", "")
                output_trans_hom.writelines(cleaned_line)
                output_trans_het.writelines(cleaned_line)
            else:
                parts = line.strip().split("\t")
                if parts[1] == parts[3]:
                    output_cis.writelines(line)
                elif ((parts[1] == "2L_JJg14_439" and parts[3] == "2L_JJg14_057") or (parts[3] == "2L_JJg14_439" and parts[1] == "2L_JJg14_057") or (parts[1] == "2R_JJg14_439" and parts[3] == "2R_JJg14_057") or (parts[3] == "2R_JJg14_439" and parts[1] == "2R_JJg14_057") or (parts[1] == "3L_JJg14_439" and parts[3] == "3L_JJg14_057") or (parts[3] == "3L_JJg14_439" and parts[1] == "3L_JJg14_057") or (parts[1] == "3R_JJg14_439" and parts[3] == "3R_JJg14_057") or (parts[3] == "3R_JJg14_439" and parts[1] == "3R_JJg14_057")):
                    cleaned_line = line.replace("_JJg14_057", "").replace("_JJg14_439", "")
                    output_trans_hom.writelines(cleaned_line)
                else:
                    cleaned_line = line.replace("_JJg14_057", "").replace("_JJg14_439", "")
                    output_trans_het.writelines(cleaned_line)

                

input_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004895_PnM1/pairs/20250403_24L004895_PnM1.filtered.pairs.gz"
output_file_cis = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004895_PnM1/pairs/20250429_24L004895_PnM1.cis.filtered.pairs"
output_file_trans_hom = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004895_PnM1/pairs/20250429_24L004895_PnM1.trans_hom.filtered.pairs"
output_file_trans_het = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004895_PnM1/pairs/20250429_24L004895_PnM1.trans_het.filtered.pairs"

#input_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250403_24L004896_PnM2.filtered.pairs.gz"
#output_file_cis = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250429_24L004896_PnM2.cis.filtered.pairs"
#output_file_trans_hom = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250429_24L004896_PnM2.trans_hom.filtered.pairs"
#output_file_trans_het = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250429_24L004896_PnM2.trans_het.filtered.pairs"

#input_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004897_PnM3/pairs/20250403_24L004897_PnM3.filtered.pairs.gz"
#output_file_cis = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004897_PnM3/pairs/20250430_24L004897_PnM3.cis.filtered.pairs"
#output_file_trans_hom = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004897_PnM3/pairs/20250430_24L004897_PnM3.trans_hom.filtered.pairs"
#output_file_trans_het = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004897_PnM3/pairs/20250430_24L004897_PnM3.trans_het.filtered.pairs"


filter_by_cis_trans(input_file,output_file_cis, output_file_trans_hom, output_file_trans_het)
```
