```python
import cooler
import pandas as pd

# Load the cooler file
#c = cooler.Cooler('/data/processing4/erikson_pipegrp/Jumana_nextflow/20250325_merged.mcool::/resolutions/10000')
resolution = 10000 # note this might be slightly slow on a laptop
                  # and could be lowered to 10kb for increased speed
clr = cooler.Cooler('/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004895_PnM1/pairs/20250325_24L004895_PnM1.pairs.dedup.cool')


# Specify the chromosomes
chrom1 = '2L_JJg14_057'
chrom2 = '2L_JJg14_439'

# Extract trans contacts between the specified chromosomes
trans_contacts = clr.matrix(as_pixels=True, balance=False).fetch((chrom1, chrom2))
print(trans_contacts.head())

# Create a DataFrame for the new cooler file
df = pd.DataFrame({
    'bin1_id': trans_contacts['bin1_id'],
    'bin2_id': trans_contacts['bin2_id'],
    'count': trans_contacts['count']
})

# Use the bins from the original cooler object
bins = clr.bins()[:]  # Extract bins from the original cooler file

# Create a new cooler file with the extracted trans contacts
cooler.create_cooler(
    '/data/processing4/erikson_pipegrp/Jumana_nextflow/20250325_24L004895_PnM1.2L.cool',
    bins=bins,
    pixels=df,
    ordered=True,
    dtypes={'count': 'int32'}  # Ensure the data type is specified for the count
)
```

```python
import sys
import gzip

def filter_pairs_by_specific_chromosomes(input_file, output_file):
    # List of chromosomes to filter
    target_chromosomes = ["2L_JJg14_439","2R_JJg14_439", "3L_JJg14_439", "3R_JJg14_439", "4_JJg14_439", "X_JJg14_439", "Y_JJg14_439", "2L_JJg14_057", "2R_JJg14_057", "3L_JJg14_057", "3R_JJg14_057", "4_JJg14_057", "X_JJg14_057", "Y_JJg14_057"]
    buffer = []
    buffer_size = 10000

    with gzip.open(input_file, 'rt') as infile, open(output_file, 'w') as outfile:
        for line in infile:
            if line.startswith("#"):  # Process header lines
                # Check if the header line starts with any of the target chromosomes
                if line.startswith("#chromsize"):
                    parts = line.strip().split(" ")
                    if parts[1] in target_chromosomes:
                        buffer.append(line)
                elif line.startswith("#samheader: @SQ"):
                    parts = line.strip().split(":")
                    parts2 = parts[2].strip().split("LN")
                    if parts2[0].strip() in target_chromosomes:
                        buffer.append(line)
                else:
                    buffer.append(line)
                if len(buffer) >= buffer_size:
                    outfile.writelines(buffer)
                    buffer = []
                continue
            parts = line.strip().split("\t")
            # Check if both parts[1] and parts[3] are in the target chromosomes list
            if parts[1] in target_chromosomes and parts[3] in target_chromosomes:
                buffer.append("\t".join(parts) + "\n")
                if len(buffer) >= buffer_size:
                    outfile.writelines(buffer)
                    buffer = []
        if buffer:
            outfile.writelines(buffer)

# Example usage
#input_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/24L004896_PnM2.pairs.gz"
#output_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250403_24L004896_PnM2.filtered.pairs"

#input_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004895_PnM1/pairs/24L004895_PnM1.pairs.gz"
#output_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004895_PnM1/pairs/20250403_24L004895_PnM1.filtered.pairs"

input_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004897_PnM3/pairs/24L004897_PnM3.pairs.gz"
output_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004897_PnM3/pairs/20250403_24L004897_PnM3.filtered.pairs"


filter_pairs_by_specific_chromosomes(input_file, output_file)
```

```python
####################
#June 6th, 2025
#Joining the trans and cis into a single file


import sys
import gzip

def filter_pairs_by_specific_chromosomes(input_file, output_file_2L, output_file_2R, output_file_3L, output_file_3R, output_file_4, output_file_X):
    # List of chromosomes to filter
    target_chromosomes = ["2L_JJg14_439","2R_JJg14_439", "3L_JJg14_439", "3R_JJg14_439", "4_JJg14_439", "X_JJg14_439", "Y_JJg14_439", "2L_JJg14_057", "2R_JJg14_057", "3L_JJg14_057", "3R_JJg14_057", "4_JJg14_057", "X_JJg14_057", "Y_JJg14_057"]
    buffer = []
    buffer_2L=[]
    buffer_2R=[]
    buffer_3L=[]
    buffer_3R=[]
    buffer_X=[]
    buffer_4=[]
    
    buffer_size = 10000

    with gzip.open(input_file, 'rt') as infile, gzip.open(output_file_2L, 'wt') as outfile_2L, gzip.open(output_file_2R, 'wt') as outfile_2R, gzip.open(output_file_3L, 'wt') as outfile_3L,  gzip.open(output_file_3R, 'wt') as outfile_3R, gzip.open(output_file_4, 'wt') as outfile_4, gzip.open(output_file_X, 'wt') as outfile_X:
        for line in infile:
            if line.startswith("#"):  # Process header lines
                # Check if the header line starts with any of the target chromosomes
                if line.startswith("#chromsize"):
                    parts = line.strip().split(" ")
                    if parts[1] in target_chromosomes:
                        #create a new chromosome trans and add it to the header
                        line = line.replace("_JJg14_057", "_trans")
                        line = line.replace("_JJg14_439", "_trans")
                        buffer.append(line)
                        #and then replace trans with nothing and add the sis chromosomes to header
                        line = line.replace("_trans", "")
                        buffer.append(line)
                elif line.startswith("#samheader: @SQ"):
                    parts = line.strip().split(":")
                    parts2 = parts[2].strip().split("LN")
                    if parts2[0].strip() in target_chromosomes:
                        line = line.replace("_JJg14_057", "_trans")
                        line = line.replace("_JJg14_439", "_trans")
                        buffer.append(line)
                        line = line.replace("_trans", "")
                        buffer.append(line)
                else:
                    buffer.append(line)

                outfile_2L.writelines(buffer)
                outfile_2R.writelines(buffer)
                outfile_3L.writelines(buffer)
                outfile_3R.writelines(buffer)
                outfile_X.writelines(buffer)
                outfile_4.writelines(buffer)
                buffer=[]
                continue
            parts = line.strip().split("\t")
            # Check if both parts[1] and parts[3] are in the target chromosomes list
            if parts[1] in target_chromosomes and parts[3] in target_chromosomes:
                
                if parts[1] == parts[3]:
                    line = line.replace("_JJg14_057", "")
                    line = line.replace("_JJg14_439", "")
                else:
                    line = line.replace("_JJg14_057", "_trans")
                    line = line.replace("_JJg14_439", "_trans")
                
                if "2L" in parts[1]  or "2L" in parts[3]:
                    buffer_2L.append(line)
                elif "2R" in parts[1] or "2R" in parts[3]:
                    buffer_2R.append(line)
                elif "3L" in parts[1] or "3L" in parts[3]:
                    buffer_3L.append(line)
                elif "3R" in parts[1] or "3R" in parts[3]:
                    buffer_3R.append(line)
                elif "X" in parts[1] or "X" in parts[3]:
                    buffer_X.append(line)
                else:
                    buffer_4.append(line)
                    
                    
                if len(buffer_2R) >= buffer_size:
                    outfile_2L.writelines(buffer_2L)
                    outfile_2R.writelines(buffer_2R)
                    outfile_3L.writelines(buffer_3L)
                    outfile_3R.writelines(buffer_3R)
                    outfile_X.writelines(buffer_X)
                    outfile_4.writelines(buffer_4)
                    buffer_2L = []
                    buffer_2R = []
                    buffer_3L = []
                    buffer_3R = []
                    buffer_4 = []
                    buffer_X = []
                    
        if buffer_2L:
            outfile_2L.writelines(buffer_2L)
            outfile_2R.writelines(buffer_2R)
            outfile_3L.writelines(buffer_3L)
            outfile_3R.writelines(buffer_3R)
            outfile_X.writelines(buffer_X)
            outfile_4.writelines(buffer_4)

# Example usage
#input_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/24L004896_PnM2.pairs.gz"
#output_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250403_24L004896_PnM2.filtered.pairs"

#input_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004895_PnM1/pairs/24L004895_PnM1.pairs.gz"
#output_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004895_PnM1/pairs/20250403_24L004895_PnM1.filtered.pairs"

#input_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004897_PnM3/pairs/24L004897_PnM3.pairs.gz"
#output_file_2L = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004897_PnM3/pairs/20250606_24L004897_PnM3.filtered.2L.pairs.gz"
#output_file_2R = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004897_PnM3/pairs/20250606_24L004897_PnM3.filtered.2R.pairs.gz"
#output_file_3L = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004897_PnM3/pairs/20250606_24L004897_PnM3.filtered.3L.pairs.gz"
#output_file_3R = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004897_PnM3/pairs/20250606_24L004897_PnM3.filtered.3R.pairs.gz"
#output_file_4 = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004897_PnM3/pairs/20250606_24L004897_PnM3.filtered.4.pairs.gz"
#output_file_X = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004897_PnM3/pairs/20250606_24L004897_PnM3.filtered.X.pairs.gz"


#input_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/24L004896_PnM2.pairs.gz"
#output_file_2L = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250606_24L004896_PnM2.filtered.2L.pairs.gz"
#output_file_2R = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250606_24L004896_PnM2.filtered.2R.pairs.gz"
#output_file_3L = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250606_24L004896_PnM2.filtered.3L.pairs.gz"
#output_file_3R = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250606_24L004896_PnM2.filtered.3R.pairs.gz"
#output_file_4 = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250606_24L004896_PnM2.filtered.4.pairs.gz"
#output_file_X = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250606_24L004896_PnM2.filtered.X.pairs.gz"

input_file = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/24L004896_PnM2.pairs.gz"
output_file_2L = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250606_24L004896_PnM2.filtered.2L.pairs.gz"
output_file_2R = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250606_24L004896_PnM2.filtered.2R.pairs.gz"
output_file_3L = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250606_24L004896_PnM2.filtered.3L.pairs.gz"
output_file_3R = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250606_24L004896_PnM2.filtered.3R.pairs.gz"
output_file_4 = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250606_24L004896_PnM2.filtered.4.pairs.gz"
output_file_X = "/data/processing4/erikson_pipegrp/Jumana_nextflow/24L004896_PnM2/pairs/20250606_24L004896_PnM2.filtered.X.pairs.gz"


filter_pairs_by_specific_chromosomes(input_file, output_file_2L, output_file_2R, output_file_3L, output_file_3R, output_file_4, output_file_X)
```
